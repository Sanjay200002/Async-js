function delay(ms,message) {
    return new Promise(resolve => {
        setTimeout(() => {
            console.log(message);
            resolve();
        }, ms);
    });
}

//5.  How does the javascript event loop handle promises differently from setTimeout?

// Solution :-

// JavaScript is single-threaded (one call stack). to handle asynchronous operation, it uses the event loop, which manage two main queues.
// 1. Macro-Task Queue Like :-
// setTimeout, setInterval, setImmediate(Node.js), DOM events, I/O callback
// -> All these are scheduled to run after the current execution finished.

// 2. Micro-Task Queue like :-
// Promises callbacks (.then, .catch, .finally)
// -> Micro-Task are always executed before moving on the next macro-Task

//          Promises vs setTimeout are handled

// setTimeout(fn, 0)
// -> Goes into the macro-Task queue.
// Executes only After:
// Curent script finishes,  the all micro-Task in the queue finish


// Promise.resolve().then(fn)  :- Goes into the micro-Task queue.
// Executes immediatelyty after the curent synchronous code but before any macro-Task

// Example :-

console.log("Start");   // first executes

setTimeout(() => {
    console.log("This is a macro-Task so last executes.")   // last Execute
}, 0);

Promise.resolve().then(()=> {
    console.log("This is micro-Task so executes before the macro task.");   // before the setTimeout  executes.
});

console.log("end");    // second executes
